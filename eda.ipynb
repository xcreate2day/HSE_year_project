```python
dfs = (df_2018, df_2019, df_2020, df_2021, df_2022)
names = ("2018", "2019", "2020", "2021", "2022")
```

```python
print("--- Количество колонок в каждом файле ---")
for name, df in zip(names, dfs):
    print(f"Файл {name}: {len(df.columns)} колонок")
```

```python
print("\nРезультаты сравнения с 2018 годом")
base_cols = set(df_2018.columns)
all_same = True

for name, df in zip(names[1:], dfs[1:]):
    current_cols = set(df.columns)
    
    if current_cols == base_cols:
        print(f"✅ {name} - колонки полностью совпадают")
    else:
        all_same = False
        missing = base_cols - current_cols
        extra = current_cols - base_cols
        
        print(f"❌ {name} - ЕСТЬ РАЗЛИЧИЯ")
        if missing:
            print(f"Не хватает колонок: {missing}")
        if extra:
            print(f"Лишние колонки: {extra}")

if all_same:
    print("\n Все 5 df имеют одинаковый набор колонок")
else:
    print("\n Наборы колонок различаются")
```
